```python
%matplotlib widget
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def bond_behavior_plots(df,experiment,kn_ks,group_type):
    plt.close("all")
    fig = plt.figure(figsize=(9,10), constrained_layout=True)
    fig.suptitle(f"Small-group {group_type} experiment {experiment}", fontsize=16)
    gs = fig.add_gridspec(4, 3)

    # X Position
    ax1 = fig.add_subplot(gs[0, 0])
    ax1.set_ylabel("Position (m)")
    ax1.plot(df["time"], df["b_x"])
    ax1.set_title("b_x")

    # Y Position
    ax2 = fig.add_subplot(gs[0, 1])
    ax2.sharex(ax1)
    ax2.plot(df["time"], df["b_y"])
    ax2.set_title("b_y")

    # Theta
    ax3 = fig.add_subplot(gs[0, 2])
    ax3.sharex(ax1)
    ax3.plot(df["time"], df["b_theta"])
    ax3.set_title("b_theta")

    # X Velocity
    ax4 = fig.add_subplot(gs[1, 0])
    ax4.sharex(ax1)
    ax4.set_ylabel("Velocity (m/s)")
    ax4.plot(df["time"], df["b_x_vel"])
    ax4.set_xlabel("Cycle Time (s)")

    # Y Velocity
    ax5 = fig.add_subplot(gs[1, 1])
    ax5.sharex(ax1)
    ax5.plot(df["time"], df["b_y_vel"])
    ax5.set_xlabel("Cycle Time (s)")

    # Omega
    ax6 = fig.add_subplot(gs[1,2])
    ax6.sharex(ax1)
    ax6.plot(df["time"], df["b_omega"])
    ax6.set_xlabel("Cycle Time (s)")

    # Contact B
    # Contact  Force
    ax7 = fig.add_subplot(gs[2,0])
    ax7.sharex(ax1)
    ax7.plot(df["time"], df["contact_b_nforce"],df["time"], df["contact_b_sforce"])
    ax7.set_xlabel("Cycle Time (s)")
    ax7.set_title("Contact B Forces")

    # Contact Moment
    ax8 = fig.add_subplot(gs[2,1])
    ax8.sharex(ax1)
    ax8.plot(df["time"], df["contact_b_moment"])
    ax8.set_xlabel("Cycle Time (s)")
    ax8.set_title("Contact B Moment")

    # Contact Bond Stress
    ax9 = fig.add_subplot(gs[2,2])
    ax9.sharex(ax1)
    ax9.plot(df["time"], df["contact_b_pb_stress"])
    ax9.set_xlabel("Cycle Time (s)")
    ax9.set_title("Contact B Bond Stress")

    # Contact A
    # Contact  Force
    ax10 = fig.add_subplot(gs[3,0])
    ax10.sharex(ax1)
    ax10.plot(df["time"], df["contact_a_nforce"],df["time"], df["contact_a_sforce"])
    ax10.set_xlabel("Cycle Time (s)")
    ax10.set_title("Contact A Forces")

    # Contact Moment
    ax11 = fig.add_subplot(gs[3,1])
    ax11.sharex(ax1)
    ax11.plot(df["time"], df["contact_a_moment"])
    ax11.set_xlabel("Cycle Time (s)")
    ax11.set_title("Contact A Moment")

    # Contact Bond Stress
    ax12 = fig.add_subplot(gs[3,2])
    ax12.sharex(ax1)
    ax12.plot(df["time"], df["contact_a_pb_stress"])
    ax12.set_xlabel("Cycle Time (s)")
    ax12.set_title("Contact A Bond Stress")

    ax1.ticklabel_format(style="sci", axis="y", scilimits=(0,0))
    ax2.ticklabel_format(style="sci", axis="y", scilimits=(0,0))
    ax3.ticklabel_format(style="sci", axis="y", scilimits=(0,0))
    ax4.ticklabel_format(style="sci", axis="y", scilimits=(0,0))
    ax5.ticklabel_format(style="sci", axis="y", scilimits=(0,0))
    ax6.ticklabel_format(style="sci", axis="y", scilimits=(0,0))
    ax7.ticklabel_format(style="sci", axis="y", scilimits=(0,0))
    ax8.ticklabel_format(style="sci", axis="y", scilimits=(0,0))
    ax9.ticklabel_format(style="sci", axis="y", scilimits=(0,0))
    ax10.ticklabel_format(style="sci", axis="y", scilimits=(0,0))
    ax11.ticklabel_format(style="sci", axis="y", scilimits=(0,0))
    ax12.ticklabel_format(style="sci", axis="y", scilimits=(0,0))

    fig2, ax = plt.subplots()
    ax.plot(df["b_x"], df["b_y"],linewidth=0.5,c="black")
    pos = ax.scatter(df["b_x"], df["b_y"],s=10,c=df["time"],cmap="coolwarm")
    fig2.colorbar(pos, ax=ax)
    ax.set_xlabel("X Position (m)")
    ax.set_ylabel("Y Position (m)")
    ax.set_title("Ball position")

    return fig, fig2
```
